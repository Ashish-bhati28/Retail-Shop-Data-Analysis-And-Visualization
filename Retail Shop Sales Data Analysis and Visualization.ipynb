**Creating MySQL Table from Data Frame**

```python
try:
    retail_shop_data.to_sql(
        name= 'sales_data',
        con=database_engine,
        if_exists='replace',
        index=False,
        dtype={'Transaction Date' : db.Date}
    )
except Exception as error:
    print(error)
else:
    print("Table is created successfully")
```

**Creating DataFrame from SQL Table**

```python
try:
    total_revenue_by_month = pd.read_sql(
        sql="""
            SELECT 
                `Transaction Month Name` as 'Month',
                sum(`Total Amount`) as 'Total Revenue'
            from 
                sales_data
            group by `Month`
            """,
            con=database_engine
            )
except Exception as error:
    print(error)
else:
    print("Query executed successfully")
```

```python
total_revenue_by_month
```

**Showing the trend line of Total Revenue By Month**

```python
sns.set_style('whitegrid')
plt.figure(figsize=(12,3))
plt.title("Total Revenue By Month")
sns.lineplot(
    data= total_revenue_by_month,
    x='Month',
    y='Total Revenue'
)
plt.show()
```

**Find the number of transaction by day**

```python
try:
    number_of_transaction_by_day = pd.read_sql(
        sql= """SELECT 
            `Transaction Day Name` AS 'Day',
            COUNT(`Transaction Day Name`) AS 'Number of Orders'
        FROM
            sales_data
        GROUP BY `Day`;
            """,
            con=database_engine
    )
except Exception as error:
    print(error)
else:
    print("Query executed seccessfully")
```

```python
number_of_transaction_by_day
```

**Compare Number of Transsaction by day using Bar plot**

```python
plt.figure(figsize=(12,3))
plt.title("Numberof Transactions by day")
sns.barplot(
    data=number_of_transaction_by_day,
    x='Day',
    y='Number of Orders',
    width=.5,
    hue='Day'
    
)
plt.show()
```

**Find Total Revenue By Category**

```python
try:
    total_revenue_by_category = pd.read_sql(
        sql="""
            select
            `Product Category` as 'Category',
            sum(`Total Amount`) as 'Total Revenue'
            from sales_data
            group by `Product Category`
            """,
                con=database_engine
    )
except Exception as error:
    print(error)
else:
    print("Query executed successfully")
```

```python
total_revenue_by_category
```

**Plot a pie chart to show percentage of Total Revenue By Category**

```python
plt.figure(figsize=(6,4))
plt.title("Total Revenue By Category")
plt.pie(
    total_revenue_by_category['Total Revenue'],
    labels=total_revenue_by_category['Category'],
    autopct="%.2f%%"
)
plt.show()
```
